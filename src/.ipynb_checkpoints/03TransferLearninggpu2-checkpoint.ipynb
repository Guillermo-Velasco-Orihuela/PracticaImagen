### **Import Libraries**

```python
from cnngpu2 import CNN, load_data
import torchvision
import torch
import torch.nn as nn
import wandb
import os
```

### PyTorch Device Setup

```python
# Configurar dispositivo
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"✅ Running on: {device}")
if device.type == "cuda":
    print(f"GPU: {torch.cuda.get_device_name(0)}")
```

### Train and Validation Paths

```python
# Paths
train_dir = os.path.join(os.path.dirname(os.getcwd()), 'dateset', 'training')
valid_dir = os.path.join(os.path.dirname(os.getcwd()), 'dateset', 'validation')
```

### Load Data and set of Hyperparameters

```python
# One set of hyperparameters
batch_size = 32
img_size = 224
learning_rate = 0.001
epochs = 40
unfrozen_layer_names = ['layer4', 'fc']

# Load data
train_loader, valid_loader, num_classes = load_data(train_dir, valid_dir, batch_size, img_size)
```

### **Models**

#### resnext101_64x4d

```python
# Load model
base_model = torchvision.models.resnext101_64x4d(weights='DEFAULT')
model = CNN(base_model, num_classes)
model.to(device)

# Freeze everything
for param in model.parameters():
    param.requires_grad = False

# Unfreeze selected layers
print("✅ Unfreezing layers:", unfrozen_layer_names)
for name, module in base_model.named_children():
    if name in unfrozen_layer_names:
        for param in module.parameters():
            param.requires_grad = True

# Ensure classifier (fc) head is unfreezed from custom model
for param in model.fc.parameters():
    param.requires_grad = True

# Print trainable param count
trainable_params = list(filter(lambda p: p.requires_grad, model.parameters()))
print(f"🔍 Trainable parameters: {sum(p.numel() for p in trainable_params)}")

if not trainable_params:
    raise RuntimeError("❌ No trainable parameters found. Check layer unfreezing logic.")

# Optimizer and loss
optimizer = torch.optim.Adam(trainable_params, lr=learning_rate)
criterion = nn.CrossEntropyLoss()

# Optional: WandB logging
wandb.init(
    project="marcosblancovigo-universidad-pontificia-comillas",
    name=f"resnext101_64x4d_model_batchsize{batch_size}_imgsize{img_size}_learningrate{learning_rate}_epoch{epochs}.pth",
    config={
        "batch_size": batch_size,
        "img_size": img_size,
        "learning_rate": learning_rate,
        "epochs": epochs,
        "unfreezed_layers": unfrozen_layer_names
    }
)

# Train
history = model.train_model(train_loader, valid_loader, optimizer, criterion, epochs=epochs, device=device)

# Log to wandb
for epoch in range(epochs):
    wandb.log({
        "epoch": epoch + 1,
        "train_loss": history["train_loss"][epoch],
        "val_loss": history["valid_loss"][epoch],
        "train_acc": history["train_accuracy"][epoch],
        "val_acc": history["valid_accuracy"][epoch],
    })


# Save model
model_path = os.path.join(os.getcwd(), f"resnext101_64x4d_model_batchsize{batch_size}_imgsize{img_size}_learningrate{learning_rate}_epoch{epochs}.pt")
torch.save(model.state_dict(), model_path)
artifact = wandb.Artifact(f"resnext101_64x4d_model_bactch{batch_size}_imgsize{img_size}_learningrate{learning_rate}_epoch{epochs}", type="model")
artifact.add_file(model_path)
wandb.log_artifact(artifact)

wandb.finish()
```
