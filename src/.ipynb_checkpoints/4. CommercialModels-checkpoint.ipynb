In this project, we aim to classify various eye diseases using a labeled image dataset. To accomplish this, we leverage the power of transfer learning by fine-tuning four state-of-the-art convolutional neural networks (CNNs) pretrained on ImageNet. The selected models vary in complexity and architecture, allowing us to compare their performance across different levels of representational power and computational cost.

The four models used are, from most complex to least complex:

- RegNetY-128GF

- ResNeXt101_64x4d

- ResNeXt50_32x4d

- ResNet50

These models have been chosen for their proven effectiveness in image classification tasks

### Hypothesis:

Given their complexity and design we hypothesize that:

- RegNetY-128GF will outperform all other models, especially in capturing fine-grained patterns in eye disease images due to its depth and architectural sophistication.

- ResNet50 is expected to have the lowest performance, as it lacks the enhanced feature representation capabilities of the ResNeXt and RegNet architectures.

- ResNeXt101_64x4d and ResNeXt50_32x4d are expected to fall in between, with the deeper ResNeXt101 likely outperforming its smaller counterpart.

### **Import Libraries**

```python
from cnn import CNN, load_data
import torchvision
import torch
import torch.nn as nn
import wandb
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report
import pandas as pd
import numpy as np
import os
import torch
import torch.nn as nn
import torchvision
import matplotlib.pyplot as plt

import os
```

### PyTorch Device Setup

```python
# Configurar dispositivo
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"✅ Running on: {device}")
if device.type == "cuda":
    print(f"GPU: {torch.cuda.get_device_name(0)}")
```

### Train and Validation Paths

```python
# Paths
train_dir = "C:\\Users\\guill\\Documents\\master\\ImagenAdne\\PracticaImagen\\dataset\\train"
valid_dir = "C:\\Users\\guill\\Documents\\master\\ImagenAdne\\PracticaImagen\\dataset\\val"
```

### Load Data and set of Hyperparameters

```python
# One set of hyperparameters
batch_size = 32
img_size = 224
learning_rate = 0.001
epochs = 40
unfrozen_layer_names = ['layer4', 'fc']

# Load data
train_loader, valid_loader, num_classes = load_data(train_dir, valid_dir, batch_size, img_size)
```

### **Models**

#### 1. resnext101_64x4d

ResNeXt101_64x4d is a deep CNN that enhances ResNet by using multiple parallel paths (called cardinality) in each block. With 101 layers and a 64x4d configuration, it captures complex features efficiently, making it powerful for tasks like medical image classification.

##### 1.1 resnext101_64x4d code

```python
# Load model
base_model = torchvision.models.resnext101_64x4d(weights='DEFAULT')
model = CNN(base_model, num_classes)
model.to(device)

# Freeze everything
for param in model.parameters():
    param.requires_grad = False

# Unfreeze selected layers
print("✅ Unfreezing layers:", unfrozen_layer_names)
for name, module in base_model.named_children():
    if name in unfrozen_layer_names:
        for param in module.parameters():
            param.requires_grad = True

# Ensure classifier (fc) head is unfreezed from custom model
for param in model.fc.parameters():
    param.requires_grad = True

# Print trainable param count
trainable_params = list(filter(lambda p: p.requires_grad, model.parameters()))
print(f"🔍 Trainable parameters: {sum(p.numel() for p in trainable_params)}")

if not trainable_params:
    raise RuntimeError("❌ No trainable parameters found. Check layer unfreezing logic.")

# Optimizer and loss
optimizer = torch.optim.Adam(trainable_params, lr=learning_rate)
criterion = nn.CrossEntropyLoss()

# Train
history = model.train_model(train_loader, valid_loader, optimizer, criterion, epochs=epochs, device=device)

# Save model
model_filename = f"resnext101_64x4d_model_batchsize{batch_size}_imgsize{img_size}_learningrate{learning_rate}_epoch{epochs}.pt"
model_path = os.path.join(os.getcwd(), model_filename)
torch.save(model.state_dict(), model_path)

# === Save Plots === #
plot_folder = os.path.join("plts", os.path.splitext(model_filename)[0])
os.makedirs(plot_folder, exist_ok=True)

def plot_metric(metric_name, train_values, val_values):
    plt.figure()
    plt.plot(range(1, epochs + 1), train_values, label='Train')
    plt.plot(range(1, epochs + 1), val_values, label='Validation')
    plt.xlabel("Epoch")
    plt.ylabel(metric_name.capitalize())
    plt.title(f"{metric_name.capitalize()} over Epochs")
    plt.legend()
    plt.grid(True)
    plt.savefig(os.path.join(plot_folder, f"{metric_name}.png"))
    plt.close()

# Plot and save all relevant metrics
plot_metric("loss", history["train_loss"], history["valid_loss"])
plot_metric("accuracy", history["train_accuracy"], history["valid_accuracy"])

# === Evaluate and Save Confusion Matrix === #
model.eval()
all_preds = []
all_labels = []

with torch.no_grad():
    for images, labels in valid_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, preds = torch.max(outputs, 1)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

# Compute confusion matrix
cm = confusion_matrix(all_labels, all_preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap='Blues', xticks_rotation='vertical')

# Save confusion matrix plot
plt.title("Confusion Matrix")
plt.savefig(os.path.join(plot_folder, "confusion_matrix.png"))
plt.close()

# Save classification report
report = classification_report(all_labels, all_preds, output_dict=True)
report_df = pd.DataFrame(report).transpose()
report_df.to_csv(os.path.join(plot_folder, "classification_report.csv"))

# Optional: also save raw confusion matrix
cm_df = pd.DataFrame(cm)
cm_df.to_csv(os.path.join(plot_folder, "confusion_matrix_raw.csv"), index=False)
```

**1.2.1 Confusion Matrix:**

The confusion matrix indicates that the model is classifying most images correctly across all four classes.
```python
# Load model
base_model = torchvision.models.regnet_y_128gf(weights='DEFAULT')
model = CNN(base_model, num_classes)
model.to(device)

# Freeze everything
for param in model.parameters():
    param.requires_grad = False

# Unfreeze selected layers
print("✅ Unfreezing layers:", unfrozen_layer_names)
for name, module in base_model.named_children():
    if name in unfrozen_layer_names:
        for param in module.parameters():
            param.requires_grad = True

# Ensure classifier (fc) head is unfreezed from custom model
for param in model.fc.parameters():
    param.requires_grad = True

# Print trainable param count
trainable_params = list(filter(lambda p: p.requires_grad, model.parameters()))
print(f"🔍 Trainable parameters: {sum(p.numel() for p in trainable_params)}")

if not trainable_params:
    raise RuntimeError("❌ No trainable parameters found. Check layer unfreezing logic.")

# Optimizer and loss
optimizer = torch.optim.Adam(trainable_params, lr=learning_rate)
criterion = nn.CrossEntropyLoss()

# Train
history = model.train_model(train_loader, valid_loader, optimizer, criterion, epochs=epochs, device=device)

# Save model
model_filename = f"regnet_y_128gf_model_batchsize{batch_size}_imgsize{img_size}_learningrate{learning_rate}_epoch{epochs}.pt"
model_path = os.path.join(os.getcwd(), model_filename)
torch.save(model.state_dict(), model_path)

# === Save Plots === #
plot_folder = os.path.join("plts", os.path.splitext(model_filename)[0])
os.makedirs(plot_folder, exist_ok=True)

def plot_metric(metric_name, train_values, val_values):
    plt.figure()
    plt.plot(range(1, epochs + 1), train_values, label='Train')
    plt.plot(range(1, epochs + 1), val_values, label='Validation')
    plt.xlabel("Epoch")
    plt.ylabel(metric_name.capitalize())
    plt.title(f"{metric_name.capitalize()} over Epochs")
    plt.legend()
    plt.grid(True)
    plt.savefig(os.path.join(plot_folder, f"{metric_name}.png"))
    plt.close()

# Plot and save all relevant metrics
plot_metric("loss", history["train_loss"], history["valid_loss"])
plot_metric("accuracy", history["train_accuracy"], history["valid_accuracy"])

# === Evaluate and Save Confusion Matrix === #
model.eval()
all_preds = []
all_labels = []

with torch.no_grad():
    for images, labels in valid_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, preds = torch.max(outputs, 1)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

# Compute confusion matrix
cm = confusion_matrix(all_labels, all_preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap='Blues', xticks_rotation='vertical')

# Save confusion matrix plot
plt.title("Confusion Matrix")
plt.savefig(os.path.join(plot_folder, "confusion_matrix.png"))
plt.close()

# Save classification report
report = classification_report(all_labels, all_preds, output_dict=True)
report_df = pd.DataFrame(report).transpose()
report_df.to_csv(os.path.join(plot_folder, "classification_report.csv"))

# Optional: also save raw confusion matrix
cm_df = pd.DataFrame(cm)
cm_df.to_csv(os.path.join(plot_folder, "confusion_matrix_raw.csv"), index=False)
```

ResNet50 is a widely used convolutional neural network known for its deep architecture and efficient performance. It consists of 50 layers and introduces residual connections, which help prevent vanishing gradients and allow the model to learn deeper representations effectively.

```python
# Load model
base_model = torchvision.models.resnet50(weights='DEFAULT')
model = CNN(base_model, num_classes)
model.to(device)

# Freeze everything
for param in model.parameters():
    param.requires_grad = False

# Unfreeze selected layers
print("✅ Unfreezing layers:", unfrozen_layer_names)
for name, module in base_model.named_children():
    if name in unfrozen_layer_names:
        for param in module.parameters():
            param.requires_grad = True

# Ensure classifier (fc) head is unfreezed from custom model
for param in model.fc.parameters():
    param.requires_grad = True

# Print trainable param count
trainable_params = list(filter(lambda p: p.requires_grad, model.parameters()))
print(f"🔍 Trainable parameters: {sum(p.numel() for p in trainable_params)}")

if not trainable_params:
    raise RuntimeError("❌ No trainable parameters found. Check layer unfreezing logic.")

# Optimizer and loss
optimizer = torch.optim.Adam(trainable_params, lr=learning_rate)
criterion = nn.CrossEntropyLoss()

# Train
history = model.train_model(train_loader, valid_loader, optimizer, criterion, epochs=epochs, device=device)

# Save model
model_filename = f"resnet50_model_batchsize{batch_size}_imgsize{img_size}_learningrate{learning_rate}_epoch{epochs}.pt"
model_path = os.path.join(os.getcwd(), model_filename)
torch.save(model.state_dict(), model_path)

# === Save Plots === #
plot_folder = os.path.join("plts", os.path.splitext(model_filename)[0])
os.makedirs(plot_folder, exist_ok=True)

def plot_metric(metric_name, train_values, val_values):
    plt.figure()
    plt.plot(range(1, epochs + 1), train_values, label='Train')
    plt.plot(range(1, epochs + 1), val_values, label='Validation')
    plt.xlabel("Epoch")
    plt.ylabel(metric_name.capitalize())
    plt.title(f"{metric_name.capitalize()} over Epochs")
    plt.legend()
    plt.grid(True)
    plt.savefig(os.path.join(plot_folder, f"{metric_name}.png"))
    plt.close()

# Plot and save all relevant metrics
plot_metric("loss", history["train_loss"], history["valid_loss"])
plot_metric("accuracy", history["train_accuracy"], history["valid_accuracy"])

# === Evaluate and Save Confusion Matrix === #
model.eval()
all_preds = []
all_labels = []

with torch.no_grad():
    for images, labels in valid_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, preds = torch.max(outputs, 1)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

# Compute confusion matrix
cm = confusion_matrix(all_labels, all_preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap='Blues', xticks_rotation='vertical')

# Save confusion matrix plot
plt.title("Confusion Matrix")
plt.savefig(os.path.join(plot_folder, "confusion_matrix.png"))
plt.close()

# Save classification report
report = classification_report(all_labels, all_preds, output_dict=True)
report_df = pd.DataFrame(report).transpose()
report_df.to_csv(os.path.join(plot_folder, "classification_report.csv"))

# Optional: also save raw confusion matrix
cm_df = pd.DataFrame(cm)
cm_df.to_csv(os.path.join(plot_folder, "confusion_matrix_raw.csv"), index=False)
```

### Conclusion

All four models performed well in classifying eye diseases, with validation accuracies consistently above 85%. **ResNeXt101_64x4d** showed the best overall performance, combining high accuracy with stable training. **ResNeXt50_32x4d** followed closely, offering a good trade-off between performance and efficiency. **RegNetY-128GF** achieved strong results but showed some inconsistency, while **ResNet50**, though simpler, delivered competitive accuracy with less stable validation loss. Overall, **ResNeXt101_64x4d** proved to be the most reliable choice.

Comparing the results with the ML and DL models, it is worth noting the great surprise of the good results of ML, reaching 90% accuracy and surpassing commercial models. However, the best models in each of the three phases obtain very similar results, which indicates that it is not necessary to make the problem very complex, but rather a good feature extraction.
