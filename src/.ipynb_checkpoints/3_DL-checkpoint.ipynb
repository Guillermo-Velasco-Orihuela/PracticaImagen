## CNN

Convolutional Neural Networks (CNNs) are a class of deep learning models designed to process data that have a grid like topology, most commonly images.
A convolutional layer slides a small learnable kernel across the input and at every spatial position performs an element wise multiplication followed by a sum, producing what is called a feature map. Because the same kernel is reused everywhere, a CNN learns features that are translation invariant and greatly reduces the number of parameters compared with a fully connected network.

Stacking several convolutional layers builds a hierarchy: early layers detect simple edges and colour contrasts, midlevel layers assemble those into textures or object parts, and deeper layers capture class specific shapes.

Pooling layers down sample the representation, enlarging the receptive field while controlling over fitting, and non linear activations such as ReLU allow the network to approximate complex functions.

At the very end, one or more fully connected layers map the learned visual representation to class logits, which are turned into probabilities through a softmax that is implicitly applied by the cross entropy loss. This notebook walks through building, training and evaluating two CNNs from scratch for multi class retinal disease detection.

Here we gather all the libraries we need:
PyTorch for model building, torchvision for dataset utilities, matplotlib for plots and NumPy for numerical work.
Then we query the runtime to see whether a CUDA‑enabled GPU is available. If it is, every tensor and operation will be moved to the GPU device so that training runs orders of magnitude faster than on the CPU.

```python
# Carga el dataset desde Hugging Face
dataset = load_dataset("nks9/NKS_EYE_DISEASE_CLASSIFICATION", split="train")

# Convierte el dataset a un DataFrame de pandas
df = dataset.to_pandas()

# Muestra las primeras filas del DataFrame
print(df.head())
```

```python
# Optional: label mapping (custom names)
label_map = {
    0: "cataract",
    1: "diabetic_retinopathy",
    2: "glaucoma",
    3: "normal"
    # Add more if needed
}

# Shuffle the dataset
df = shuffle(df, random_state=42)

# Get all unique labels
unique_labels = df['label'].unique()

# Loop through each class
for label in unique_labels:
    label_str = label_map[label] if label in label_map else str(label)

    # Subset DataFrame for this class
    class_df = df[df['label'] == label]

    # Split: first 100 for validation, rest for training
    val_df = class_df.iloc[:200]
    train_df = class_df.iloc[200:]

    # Create directories
    os.makedirs(f"train/{label_str}", exist_ok=True)
    os.makedirs(f"val/{label_str}", exist_ok=True)

    # Save training images
    for idx, row in train_df.iterrows():
        image = Image.open(io.BytesIO(row['image']['bytes']))
        if image.mode != "RGB":
            image = image.convert("RGB")
        filename = f"image_{idx}.jpg"
        image.save(os.path.join(f"train/{label_str}", filename))

    # Save validation images
    for idx, row in val_df.iterrows():
        image = Image.open(io.BytesIO(row['image']['bytes']))
        if image.mode != "RGB":
            image = image.convert("RGB")
        filename = f"image_{idx}.jpg"
        image.save(os.path.join(f"val/{label_str}", filename))
```

```python
BATCH_SIZE   = 32
NUM_WORKERS  = 4
NUM_EPOCHS   = 25
LR           = 1e-3
IMG_SIZE     = 224

train_transforms = transforms.Compose([
    transforms.RandomResizedCrop(IMG_SIZE, scale=(0.8, 1.0)),
    transforms.RandomRotation(10),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std =[0.229, 0.224, 0.225]),
])

val_transforms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std =[0.229, 0.224, 0.225]),
])
```

In the code above, we declare a set of global hyperparameters that determine the training regime: the mini‑batch size, the learning rate, the number of epochs, the image resolution and the amount of parallelism in the data loaders.
 Right after that, we define two transformation pipelines. The training pipeline performs random crops, flips, small rotations and mild colour jitter to synthetically enlarge the dataset, then converts the image to a tensor and normalises it with ImageNet statistics. The validation pipeline is deterministic: it only resizes, converts and normalises, ensuring that performance is measured on unaugmented data.

```python
train_dir = "/content/train"
val_dir   = "/content/val"

train_ds = datasets.ImageFolder(train_dir, transform=train_transforms)
val_ds   = datasets.ImageFolder(val_dir,   transform=val_transforms)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE,
                          shuffle=True,  num_workers=NUM_WORKERS, pin_memory=True)

val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE,
                          shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

class_names = train_ds.classes
print("Clases:", class_names)
```

ImageFolder scans each subfolder name to infer the class index, builds an internal list of sample paths and their labels, and applies the transform pipeline on the fly. DataLoader wraps the dataset to deliver shuffled mini‑batches, optionally in parallel worker processes, and pins the memory so that the GPU can ingest it efficiently.

```python
class MiniCNN(nn.Module):
    """
    3 conv blocks + FC
    """
    def __init__(self, num_classes):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1),  nn.ReLU(),
            nn.MaxPool2d(2),                 # 112×112

            nn.Conv2d(32, 64, 3, padding=1), nn.ReLU(),
            nn.MaxPool2d(2),                 # 56×56

            nn.Conv2d(64, 128, 3, padding=1),nn.ReLU(),
            nn.MaxPool2d(2),                 # 28×28
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128*28*28, 256), nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x
```

 Here we define a network that introduces the essential building blocks. Three convolutional stages progressively increase the channel depth while halving the spatial resolution via max‑pooling. The resulting activation tensor is flattened and passed through a fully connected layer with a dropout for regularisation before the final linear layer outputs one logit per class. All hidden layers use ReLU activations, striking a good balance between gradient flow and computational cost.

```python
class DeepCNN(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        def block(in_ch, out_ch):
            return nn.Sequential(
                nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(),
                nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(),
                nn.MaxPool2d(2)
            )
        self.features = nn.Sequential(
            block(3,   32),   # 112×112
            block(32,  64),   # 56×56
            block(64, 128),   # 28×28
            block(128,256),   # 14×14
        )
        self.avgpool = nn.AdaptiveAvgPool2d((1,1))
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(256, 256), nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.avgpool(x)
        x = self.classifier(x)
        return x
```

Now, we define a more complex architecture in which each stage contains two consecutive convolution, batch, normalisation, ReLU triplets, which double the channel count and are followed by pooling.
Batch normalisation accelerates convergence and adds a regularising effect. An adaptive global average pool squeezes the final feature maps down to one spatial position so that the subsequent fully connected classifier is independent of the original image size. Compared with previous one, the deeper model can capture finer grained patterns at the cost of additional parameters.

```python
def train_epoch(model, loader, criterion, optimizer):
    model.train()
    running_loss, correct, total = 0.0, 0, 0
    for inputs, targets in loader:
        inputs, targets = inputs.to(device), targets.to(device)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * inputs.size(0)
        _, preds = outputs.max(1)
        correct += preds.eq(targets).sum().item()
        total   += targets.size(0)
    epoch_loss = running_loss / total
    epoch_acc  = correct / total
    return epoch_loss, epoch_acc

@torch.no_grad()
def eval_epoch(model, loader, criterion):
    model.eval()
    running_loss, correct, total = 0.0, 0, 0
    for inputs, targets in loader:
        inputs, targets = inputs.to(device), targets.to(device)
        outputs = model(inputs)
        loss = criterion(outputs, targets)

        running_loss += loss.item() * inputs.size(0)
        _, preds = outputs.max(1)
        correct += preds.eq(targets).sum().item()
        total   += targets.size(0)
    epoch_loss = running_loss / total
    epoch_acc  = correct / total
    return epoch_loss, epoch_acc
```

We define here functions to  encapsulate the logic for an epoch of training and one of validation. They handle gradient management, loss computation, metric accumulation and return the average loss and accuracy. The evaluator runs with torch.no_grad() so that intermediate activations are not stored, saving memory and speeding up validation

```python
def train_model(model, epochs=NUM_EPOCHS, lr=LR):
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=lr)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'min', patience=3)

    history = {"train_loss":[], "val_loss":[], "train_acc":[], "val_acc":[]}

    best_val_acc = 0.0
    for epoch in range(1, epochs+1):
        t0 = time.time()
        tr_loss, tr_acc = train_epoch(model, train_loader, criterion, optimizer)
        val_loss, val_acc = eval_epoch(model, val_loader, criterion)
        scheduler.step(val_loss)

        history["train_loss"].append(tr_loss)
        history["val_loss"].append(val_loss)
        history["train_acc"].append(tr_acc)
        history["val_acc"].append(val_acc)

        if val_acc > best_val_acc:   # guarda el mejor
            best_val_acc = val_acc
            torch.save(model.state_dict(), "best_model.pth")

        print(f"[{epoch:02}/{epochs}] "
              f"train_loss={tr_loss:.4f} val_loss={val_loss:.4f} "
              f"train_acc={tr_acc:.3f} val_acc={val_acc:.3f} "
              f"lr={optimizer.param_groups[0]['lr']:.2e}  "
              f"{(time.time()-t0):.1f}s")
    return history
```

The train_model routine orchestrates the process across multiple epochs.
It instantiates the optimiser (Adam) and a scheduler (ReduceLROnPlateau) that automatically lowers the learning rate whenever the validation loss stops improving. At the end of each epoch the routine prints a report, records the metrics for later plotting and checkpoints the network that achieved the highest validation accuracy so far.

### Train Mini CNN

```python

mini = MiniCNN(num_classes=len(class_names)).to(device)
hist_mini = train_model(mini)
```

```python
# --- Celda de código ---
def evaluate_model(model):
    model.eval()
    y_true, y_pred = [], []
    with torch.no_grad():
        for inputs, targets in val_loader:
            inputs = inputs.to(device)
            outputs = model(inputs)
            _, preds = outputs.max(1)
            y_true.extend(targets.numpy())
            y_pred.extend(preds.cpu().numpy())

    print(classification_report(y_true, y_pred, target_names=class_names, digits=4))

    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6,5))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=class_names, yticklabels=class_names)
    plt.ylabel("True"); plt.xlabel("Predicted"); plt.title("Confusion Matrix")
    plt.show()


mini_best = MiniCNN(num_classes=len(class_names)).to(device)
mini_best.load_state_dict(torch.load("best_model.pth"))
print("==== EVALUATION OF THE MODEL ====")
evaluate_model(mini_best)


```

### Train Deep CNN

```python
deep = DeepCNN(num_classes=len(class_names)).to(device)
hist_deep = train_model(deep)
```

```python
deep_best = DeepCNN(len(class_names)).to(device)
deep_best.load_state_dict(torch.load("best_model.pth"))
print("==== EVALUATION OF THE MODEL ====")
evaluate_model(deep_best)


```

```python

def plot_history(history, title):
    epochs = range(1, len(history["train_loss"])+1)
    plt.figure(figsize=(12,4))
    plt.subplot(1,2,1)
    plt.plot(epochs, history["train_loss"], label="train")
    plt.plot(epochs, history["val_loss"],   label="val")
    plt.title(f"{title} – Pérdida")
    plt.xlabel("epoch"); plt.ylabel("loss"); plt.legend()

    plt.subplot(1,2,2)
    plt.plot(epochs, history["train_acc"], label="train")
    plt.plot(epochs, history["val_acc"],   label="val")
    plt.title(f"{title} – Precisión")
    plt.xlabel("epoch"); plt.ylabel("accuracy"); plt.legend()
    plt.show()

plot_history(hist_mini, "MiniCNN")
plot_history(hist_deep, "DeepCNN")
```

In the first run the model is the MiniCNN with a learning rate of 0.001. Despite its simplicity, this compact network achieves impressive accuracy around 88% and balanced performance across classes. The classification report shows F1-scores above 0.90 for cataract and diabetic retinopathy, suggesting that the model successfully learns to distinguish between these conditions. However, performance drops on glaucoma and normal images, where confusion is more frequent. This is also reflected in the confusion matrix: while cataract and diabetic retinopathy are mostly correctly classified, there are consistent misclassifications between glaucoma and normal classes. The training and validation curves for MiniCNN indicate a stable learning pattern both loss and accuracy improve steadily over epochs, suggesting the model trains smoothly with little overfitting.

We then move on to the DeepCNN model with the same learning rate 0.001. This deeper architecture is expected to extract richer features, and while it maintains a similar global accuracy of around 88%, its classification report shows a slight shift in behavior. While cataract and diabetic retinopathy still perform well, glaucoma now suffers more, with lower precision and recall. The confusion matrix highlights this, showing substantial misclassification of glaucoma as normal and vice versa. Notably, the learning curves for DeepCNN fluctuate more dramatically especially the validation loss, which spikes at several epochs. This suggests that, despite the network's capacity, it may be more sensitive to noise or outliers in the data and might benefit from more aggressive regularization or better tuning. Overall, while DeepCNN shows potential, it also introduces more variability, and in this specific configuration does not outperform MiniCNN decisively.

To further explore the models' capacity and sensitivity to learning dynamics, the next step is to increase the learning rate to 0.01 and repeat the experiments. The goal is to observe whether faster learning translates into better performance or if it destabilizes training, especially for the more complex DeepCNN. By comparing these results, we aim to determine the optimal configuration for each architecture and draw conclusions about the trade-off between simplicity, depth, and training dynamics.

### Train Mini CNN with different LR

```python
deep = DeepCNN(num_classes=len(class_names)).to(device)
hist_deep = train_model(deep,lr=1e-2)
```

```python
deep_best = DeepCNN(len(class_names)).to(device)
deep_best.load_state_dict(torch.load("best_model.pth"))
print("==== DeepCNN ====")
evaluate_model(deep_best)
```

```python
plot_history(hist_mini, "MiniCNN")
plot_history(hist_deep, "DeepCNN")
```

The MiniCNN struggles significantly when trained with the higher learning rate. The classification report reveals catastrophic failure, with zero precision, recall, and F1-score for cataract and glaucoma, and nearly perfect recall only for diabetic retinopathy. The confusion matrix supports this: nearly all samples are misclassified as diabetic retinopathy, with virtually no useful discrimination. This behavior is likely due to gradient instability or divergence early in training. The learning curves confirm this loss drops sharply to zero in just a couple of epochs, and accuracy flatlines below 30%. The model appears to have latched onto a degenerate strategy, memorizing one class and ignoring the rest. This confirms that MiniCNN, being a smaller and less regularized network, is highly sensitive to aggressive learning rates and quickly destabilizes.

The DeepCNN, when subjected to the same learning rate of 0.01, exhibits a much more resilient behavior. While the classification report shows a drop in overall accuracy down to 56% compared to its performance at 0.001, the network still manages to assign meaningful predictions to all classes. Precision and recall values, although lower, remain relatively balanced especially for cataract and diabetic retinopathy. The confusion matrix shows the model often confuses glaucoma with normal and sometimes swaps diabetic retinopathy with cataract, but it no longer collapses into a single-class predictor. The learning curves reflect a more chaotic trajectory, with validation loss and accuracy bouncing during early epochs, but gradually stabilizing in the second half of training. These results suggest that while the DeepCNN can survive a high learning rate, its performance suffers unless the optimization dynamics are better controlled.

Comparing the loss and accuracy curves of both models reveals a clear gap in robustness. MiniCNN fails almost immediately, its training loss crashes and validation accuracy stalls. DeepCNN, on the other hand, undergoes more turbulent early training but eventually converges to a usable solution. Accuracy for DeepCNN improves steadily after the 10th epoch, and its final training and validation scores track each other closely. While not optimal, this resilience highlights the value of depth, normalization, and dropout in taming unstable learning regimes. The experiment clearly shows that deeper networks not only improve accuracy but also tolerate wider hyperparameter ranges.

Raising the learning rate exposes a stark contrast between the architectures. MiniCNN lacks the stability mechanisms needed to learn effectively under these conditions, while DeepCNN’s design allows it to recover and generalize, albeit less accurately than before. The takeaway is that deeper models offer greater robustness, but both architectures require careful tuning to achieve peak performance. These findings emphasize the importance of matching learning rate schedules to model capacity and complexity.
